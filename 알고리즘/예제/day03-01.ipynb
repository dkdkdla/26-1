## 3) DataFrame

```python
price = {
    '2020': [2500, 1500, 3000],
    '2021': [2700, 1800, 3600],
    '2022': [3000, 2400, 4200],
    '2023': [3500, 3000, 5600] 
}
#price = [
#    [3500, 3000, 2700, 2500],
#    [3000, 2400, 1800, 1500],
#    [5600, 4200, 3600, 3000]
#]
   
index = [ '사과', '딸기', '키위' ]
columns = ['2023', '2022', '2021', '2020']
```

```python
df = pd.DataFrame(price, index=index, columns=columns)
df
```
