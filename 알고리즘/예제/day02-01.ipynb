## 1) 기본자료형

```python
d1 = 34 # int 자료형
d2 = 3.4
d3 = '3'
d4 = 8 + 3j
d5 = True
print(d1)
print(type(d1))
print(type(d2))
print(type(d3))
print(type(d4))
print(type(d5))
```

## 2) bytes 와 str 비교

```python
b1 = 0b0111
b2 = 'abcd'
b3 = b'abcd'
print(b1)
print(bin(b1))
print(b2)
# print(bin(b2))
print(b3)
print(type(b2))
print(type(b3))

st1=bytes.decode(b3)
by1=str.encode(b2)

print(st1)
print(by1)
```

## 4) 논리연산자

```python
x = True
a = (0>7) or (6>9)
b = (2>3) and (1>0)
c = not (1>0)
d = (4>3) and x

print(a)
print(b)
print(c)
print(d)
```

논리연산자  or의 단락 평가(Short-circuit evaluation) :  왼쪽 값이 거짓이면 오른쪽 값 </br>
논리연산자  and의 단락 평가(Short-circuit evaluation) : 왼쪽 값이 참이면 오른쪽 값

```python
print(0 or 0)
print(0 or 1)
print(0 or 2) 
print(1 and 0)
print(1 and 1)
print(1 and 2)
```

## 7) 문자열 인터닝(String Interning)

```python
a = "String"
b = "String"
print(id(a))
print(id(b))
print(a is b)
b += "1"
b = b[:-1]
print(b)
print(a is b)
```

```python
a = 'hello'
c = "he" + "llo"
d = "".join(["h", "e", "l", "l", "o"])
print(a is c)
print(c is d)  
print(c == d)  
```
