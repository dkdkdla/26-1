## 2) String

```python
a = 'Python 3.6.4'
b = "It's you"
c = 'It\'s you'
d = "alpha\nbeta"

print(a)
print(b)
print(c)
print(d)
```

```python
a = "al"
b = "pha"
c = "alphabet"

print(a + b)
print(a * 2)
print(a[-1])
print(b[0:1])
print(c[0:-1:2])
```

```python
a = "%d %f %e" % (10, 10, 10)
b = "%o %x" % (8, 16)
c = "%c %s" % ('A', "AAA")

print(a)
print(b)
print(c)
```

```python
a = "abcd"
b = "ABCD"

print(a.upper())
print(a.lower())
```

```python
a = "   l   aa   l   "

print(a.strip())
print(a.rstrip())
print(a.lstrip())
```

```python
a = "Time is an illusion."

print(a.find('x'))
print(a.index('u'))
print(a.count('i'))
```

```python
a = "Time is an illusion."
b = "/"

print(a.split())
print(a.replace("Time", "Today"))
print(b.join(a))
print(b.startswith('/'))
```

## 4) Dictionary

```python
a = {"one": "하나", "two": "둘", "three": "셋"}
b = dict.fromkeys(["one", "two", "three"], "알 수 없음")

print(a)
print(b)
```

```python
a = {"one": "하나", "two": "둘", "three": "셋"}

print(a["one"])
print(a.get("two"))
print(a.get("four", "넷"))
```

```python
a = {"one": "하나", "two": "둘", "three": "셋"}

a["four"] = "넷"
# a.setdefault("four", "넷")
print(a["four"])

a.update({"five": "다섯", "six": "여섯"})
print(a)
```

```python
a = {"one": "하나", "two": "둘", "three": "셋"}

a["three"] = "둘"
a.update(two="셋")

print(a)
```

```python
a = {"one": "하나", "two": "둘", "three": "셋"}
b = {"five": "다섯", "six": "여섯"}

a.update(b)

print(a)
print(b)
```

```python
a = {"one": "하나", "two": "둘", "three": "셋"}
del a["three"]
# a.pop("three")
# a.popitem()
print(a)

a.clear()
print(a)
```

```python
a = {"one": "하나", "two": "둘", "three": "셋"}

print(a.items())
print(a.keys())
print(a.values())
print("one" in a)
print("four" not in a)
```

## 5) Set

```python
a = {1, 2, 3}
b = {1, 3, 5, 7, 9}

print(a)
print(b)
```

```python
a = {1, 2, 3}
b = {1, 3, 5, 7, 9}

a.add(4)
b.add(11)

print(a)
print(b)
```

```python
a = {1, 2, 3}
b = {1, 3, 5, 7, 9}

a.discard(3)
b.discard(7)

print(a)
print(b)
```

```python
L = [1, 2, 7, 2, 3]

a = set(L)

print(a)
```

```python
a = {1, 2, 3}
b = {1, 3, 5, 7, 9}

c = a | b
d = a.union(b)

print(c)
print(d)
```

```python
a = {1, 2, 3}
b = {1, 3, 5, 7, 9}

c = a & b
d = a.intersection(b)

print(c)
print(d)
```

```python
a = {1, 2, 3}
b = {1, 3, 5, 7, 9}

c = a - b
d = b - a

print(c)
print(d)
```
