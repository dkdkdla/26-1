```python
n = int(input())
print(n)
```

```python
result = 0
for i in range(1, n + 1):
    result += i
    print(result)

print(result)
```

```python
cnt = 1
result = 0
while cnt <= n:
    result += cnt
    cnt += 1
    print(result)

print(result)
```
