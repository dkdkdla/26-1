```python
def recurse(count):
    if count == 0:
        print('.')
    else:
        print('*')
        recurse(count-1)
```

```python
recurse(4)
```

```python
def factorial(n):
    print(n)
    if n == 1:
        return 1
    else:
        return n * factorial(n-1)
```

```python
print(factorial(4))
```
