```python
def isStackFull() :
    global SIZE,stack,top
    if (top >= SIZE-1) :
        return True
    else : 
        return False
    
    
```

```python
stack = ["다현", "정연", "쯔위", "사나", "지효"]
top = 4
print(isStackFull())
```

```python
def push(data) :
    global SIZE,stack,top
    if (isStackFull()) :
        print('Stack Full')
        return
    top += 1
    stack[top] = data
    
```

```python
stack = ["다현", "정연", "쯔위", "사나", None]
top = 3
print(stack,top)
push('지효')
print(stack,top)
```

```python
def isStackEmpty() :
    global SIZE,stack,top
    if (top == -1) :
        return True
    else :
        return False
        
```

```python
print(stack)
print(isStackEmpty())
```

```python
def pop() :
    global SIZE,stack,top
    if (isStackEmpty()) :
        print('Stack is Empty')
        return None
    data = stack[top]
    stack[top] = None
    top -= 1
    return data
```

```python
stack = ["다현", "정연", "쯔위", "사나", None]
top = 3
```

```python
ret_data = pop()
print(ret_data)
print(stack)
```
