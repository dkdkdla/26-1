## 1) 사용자 정의 함수

```python
def add(a, b):
    return a + b


print(add(3, 4))
print(add([1, 2, 3], [4, 5, 6]))
print(add("YUN", "DAEHEE"))
```

```python
def add(a, b=10):
    return a + b


print(add(1))
print(add(1, 2))
```

```python
def add(*num):
    result = 0

    for i in num:
        result += i

    return result


print(add(1, 2, 3))
```

```python
def calc(a, b):
    return a + b, a - b, a * b, a / b


print(calc(3, 1))
```

```python
def add(a, b):
    return a + b


def func(x, y, z):
    return x(y, z)


plus = add

print(plus(3, 4))
print(func(plus, 3, 4))
```

## 2) lambda 함수

```python
f = lambda x, y: x + y

print(f(1, 2))
print(f([1, 2], [3, 4]))
print(f("YUN", "DAEHEE"))
```

```python
L = [[9, 1], [8, 2], [7, 3], [6, 4]]

max1 = max(L, key=lambda x: x[0])
max2 = max(L, key=lambda x: x[1])
min1 = min(L, key=lambda x: x[0])
min2 = min(L, key=lambda x: x[1])

print(max1)
print(max2)
print(min1)
print(min2)
```
